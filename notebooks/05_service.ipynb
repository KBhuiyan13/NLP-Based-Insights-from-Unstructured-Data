Loading the service

```python
import os, sys
sys.path.insert(0, os.path.abspath("../src"))
from insights_service import InsightService

svc = InsightService()
svc.overview()
```

Searching with filters

```python
import pandas as pd
res = svc.search("coffee pods leaking in the machine", k=5, sentiment="negative")
pd.DataFrame(res)[["score", "year", "topic", "summary", "similarity"]]
```

Summarizing question

```python
out = svc.summarize("why do customers complain about dog treats?", sentiment="negative")
for s in out["summary"]:
    print(f"• {s['sentence']}  (review {s['review_id']})")
```

Summarizing a whole topic

```python
coffee_id = next(t["id"] for t in svc.topics if "Coffee" in t["label"])
for s in svc.summarize(topic_id=coffee_id, sentiment="negative")["summary"]:
    print("•", s["sentence"])
```
